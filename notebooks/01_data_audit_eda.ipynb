# Análise Exploratória dos dados

**Base:** `precos_carros_brasil.csv`

## Parte 1 – Importação das bibliotecas

```python
import pandas as pd
import warnings

warnings.filterwarnings('ignore')
```

---
### 1.a – Carregar a base

```python
dados = pd.read_csv('precos_carros_brasil.csv')

print('Dimensões (linhas, colunas):', dados.shape)
print('\nPrimeiras linhas:')
display(dados.head())
```

### 1.b – Valores faltantes e tratativa

```python
# Quantidade de valores faltantes por coluna
valores_faltantes = dados.isna().sum()
print('Valores faltantes por coluna:')
print(valores_faltantes)
```

```python
# Tratativa: excluir qualquer linha que contenha valor faltante (nesta base, são linhas
# inteiramente vazias, sem uso para modelagem ou estatísticas).
dados = dados.dropna(how='all')
print('\nApós remoção das linhas com valores faltantes — dimensões:', dados.shape)
print('Valores faltantes restantes:', dados.isna().sum().sum())
```

### 1.c – Dados duplicados

```python
qtd_duplicadas = dados.duplicated().sum()
print('Quantidade de linhas completamente duplicadas:', qtd_duplicadas)
print('\nLinhas duplicadas:')
dados[dados.duplicated(keep=False)]
```

### 1.d – Colunas numéricas e categóricas – estatística descritiva

```python
colunas_numericas = dados.select_dtypes(include='number').columns.tolist()
colunas_categoricas = dados.select_dtypes(include=['object', 'category']).columns.tolist()

print('Colunas numéricas:', colunas_numericas)
print('\nColunas categóricas (tipo object):', colunas_categoricas)
```

```python
print('Resumo das variáveis numéricas:')
dados[colunas_numericas].describe()
```

```python
print('Resumo das variáveis categóricas:')
dados[colunas_categoricas].describe(include='object')
```

### 1.e – Contagem por modelo e por marca

```python
print('Contagem por marca (brand):')
contagem_brand = dados['brand'].value_counts()
display(contagem_brand)

print('\nContagem por modelo (model):')
contagem_model = dados['model'].value_counts()
display(contagem_model)
```
